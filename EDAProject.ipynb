```python
df['Date of Admission'] = pd.to_datetime(df['Date of Admission'])
df['Discharge Date'] = pd.to_datetime(df['Discharge Date'])
```

# Fature Engineering

```python
df.head(2)
```

```python
df['Age'].unique()
```

```python
def condition(age):

  if age <= 30:
    return 'Adult'

  elif age<=60:
    return 'Senior'

  else:
    return 'Senior Citizen'
```

```python
df['Age_Status'] = df['Age'].apply(condition)
```

```python
df.head()
```

#Plot the Chart

```python
df.columns
```

```python
df['Medical Condition'].unique()
```

```python
# Q1. Medical Conditions of Peoples.

sns.countplot(x = 'Medical Condition', data=df, palette='rocket', hue='Age_Status')
plt.title('Medical Conditions of Peoples.')
plt.tight_layout()
plt.show()
```

```python
# Q2 Monthly wise patient Count.

df['Month'] = df['Date of Admission'].dt.month_name()
```

```python
df.head(3)
```

```python
gr = df.groupby('Month')['Date of Admission'].count()

gr.plot(kind='line', marker='o')
plt.title('Monthly wise patient Count')
plt.tight_layout()
plt.grid()
plt.show()
```

```python
# Q3 : Admission type wise Billing Amount

df.head(1)
```

```python
sns.barplot(x='Admission Type', y = 'Billing Amount', data=df, hue='Medical Condition', palette='tab10')
plt.title('Medical Condition')
plt.tight_layout()
plt.grid()
plt.show()
```

```python
# Q4. Gender Patients Distribution

pie = df.groupby('Gender')['Gender'].count()
pie.plot(kind='pie', autopct = '%1.1f%%')
plt.title('Gender Patients Distribution')
plt.ylabel('')
plt.show()
```

```python
# Q5. age_status wise patiens

age_status = df.groupby('Age_Status')['Age_Status'].count()
plt.title('age_status wise patiens')
age_status.plot(kind='pie',  autopct = '%1.1f%%')
plt.show()
```
